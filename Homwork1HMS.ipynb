```python
#Bai 6
a = int(input("Nhập vào cạnh a: "))
b = int(input("Nhập vào cạnh b: "))
c = int(input("Nhập vào cạnh c: "))
if a+b>c and a+c>b and b+c>a:
    p=(a+b+c)/2
    s=(p*(p-a)*(p-b)*(p-c))**0.5
    print(f"Diện tích tam giác là: {s:.1f}")  
else:
    print("Ba cạnh a, b, c không tạo thành một tam giác.")
    
```
